Datatypes

```python
# Creating a series of car types
cars = pd.Series(["BMW", "Toyota", "Honda"])
cars
```

```python
# Creating a series of colours
colours = pd.Series(["Blue", "Red", "White"])
colours
```

```python
# Creating a DataFrame of cars and colours
car_data = pd.DataFrame({"Car type": cars, 
                         "Colour": colours})
car_data
```
